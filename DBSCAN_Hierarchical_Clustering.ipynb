```python
#Import libraries
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import DBSCAN
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
```

```python
#Create Dataset
X, y = make_blobs(
    n_samples=200,
    centers=3,
    random_state=42
)

plt.scatter(X[:, 0], X[:, 1])
plt.title("Original Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Apply DBSCAN
dbscan = DBSCAN(eps=0.5, min_samples=5)

labels1 = dbscan.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels1)
plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Apply Hierarchical Clustering
hc = AgglomerativeClustering(n_clusters=3)

labels2 = hc.fit_predict(X)

plt.scatter(X[:, 0], X[:, 1], c=labels2)
plt.title("Hierarchical Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
#Compare the Number of Clusters
dbscan_clusters = len(set(labels1)) - (1 if -1 in labels1 else 0)
hierarchical_clusters = len(set(labels2))

print("DBSCAN clusters:", dbscan_clusters)
print("Hierarchical clusters:", hierarchical_clusters)
```

```python
#Compare Using Silhouette Score
dbscan_score = silhouette_score(X, labels1)
hierarchical_score = silhouette_score(X, labels2)

print("DBSCAN Silhouette Score:", round(dbscan_score, 3))
print("Hierarchical Silhouette Score:", round(hierarchical_score, 3))
```

```python
#Final Comparison Graph
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=labels1)
plt.title("DBSCAN")

plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=labels2)
plt.title("Hierarchical Clustering")

plt.show()
```
